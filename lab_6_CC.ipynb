### Q1: Sensor Readings Vectorization & Boolean Indexing
- a) Add 2°C to all elements using vectorization.
- b) Convert Celsius to Fahrenheit using $F = \frac{9}{5}C + 32$.
- c) Identify readings > 32°C using Boolean indexing.
- d) Count how many readings exceed 32°C.
- e) Efficiency explanation.

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("Original Readings:", temperature)

calibrated_temp = temperature + 2
print("a. Calibrated Readings (+2°C):", calibrated_temp)

temp_fahrenheit = (9 / 5) * temperature + 32
print("b. Fahrenheit Readings:", temp_fahrenheit)

greater_than_32 = temperature[temperature > 32]
print("c. Readings > 32°C:", greater_than_32)

count_greater_32 = np.sum(temperature > 32)
print("d. Count of readings > 32°C:", count_greater_32)

explanation_q1 = """
e. Efficiency Explanation:
   - Vectorization offloads repetitive element-wise operations to compiled, optimized C/Fortran
     loops inside NumPy, eliminating Python's dynamic type-checking and interpreter loop overhead.
   - Boolean Indexing creates contiguous masks directly in low-level memory, enabling ultra-fast filtering
     and aggregation without the overhead of maintaining conditional branching logic in Python.
"""
print(explanation_q1)
```

### Q2: Matrix Aggregation on Daily Steps Data
- a) Total steps recorded.
- b) Mean number of steps.
- c) Max and min values.
- d) Total steps per day (`axis=0`).
- e) Total steps per user (`axis=1`).
- f) User/day position of max steps using `argmax()`.

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Steps Matrix (Users x Days):\n", steps)

print("\na. Total steps recorded:", np.sum(steps))

print("b. Mean steps:", np.mean(steps))

print(f"c. Max steps: {np.max(steps)}, Min steps: {np.min(steps)}")

print("d. Total steps for each day (axis=0):", np.sum(steps, axis=0))

print("e. Total steps for each user (axis=1):", np.sum(steps, axis=1))

flat_max_idx = np.argmax(steps)
user_idx, day_idx = np.unravel_index(flat_max_idx, steps.shape)
print(f"f. Max steps position -> User (row): {user_idx}, Day (column): {day_idx} (Value: {steps[user_idx, day_idx]})")
```

### Q3: Slicing, Views vs Copies, Reshaping, and Flattening

```python
original = np.array([1, 2, 3, 4, 5, 6])
print("a. Original Array:", original)

subset = original[1:4]

subset[0] = 999
print("\nc. After modifying subset[0] = 999:")
print("   Original:", original)
print("   Subset (View):", subset)

original_fresh = np.array([1, 2, 3, 4, 5, 6])
copied_subset = original_fresh[1:4].copy()
copied_subset[0] = 500
print("\nd. After modifying copied_subset[0] = 500:")
print("   Original:", original_fresh)
print("   Copied Subset:", copied_subset)

mat3x4 = np.arange(1, 13).reshape(3, 4)
print("\ne. 3x4 Matrix:\n", mat3x4)

print("\nf. Extractions from 3x4 matrix:")
print("   First row:", mat3x4[0, :])
print("   Last row:", mat3x4[-1, :])
print("   Second column:", mat3x4[:, 1])
print("   Rows 1-2 and Columns 2-3:\n", mat3x4[0:2, 1:3])

flat_arr = mat3x4.flatten()
ravel_arr = mat3x4.ravel()

ravel_arr[0] = 111
print("\nh. After modifying ravel()[0] = 111:")
print("   Original Matrix (Affected - ravel returns a view):\n", mat3x4)

flat_arr[1] = 222
print("\ni. After modifying flatten()[1] = 222:")
print("   Original Matrix (Unaffected - flatten returns a deep copy):\n", mat3x4)

print("\nj. Matrix Attributes:")
print("   Shape:", mat3x4.shape)
print("   Ndim:", mat3x4.ndim)
print("   Size:", mat3x4.size)
print("   Dtype:", mat3x4.dtype)
```

### Q4: Cognitive Assistive System (Linear Regression using OLS)
Implement $\beta = (X^T X)^{-1} X^T y$ and predict for a new user.

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("a. Shape of X:", X.shape, "| Dimensions (ndim):", X.ndim)

X_T = X.T
print("\nb. Transpose of X (X.T):\n", X_T)
print("   Explanation: The transpose flips the matrix dimensions from (4 samples x 3 features) to (3 features x 4 samples), grouping each individual feature across all users into rows.")

XTX = X_T @ X
print("\nc. Matrix product (X.T @ X):\n", XTX)

XTX_inv = np.linalg.inv(XTX)
print("\nd. Inverse (X.T @ X)^(-1):\n", XTX_inv)

beta = XTX_inv @ X_T @ y
print("\ne. Calculated OLS Beta Coefficients:", beta)

print("\nf. Feature Coefficient Interpretation:")
print(f"   - Sleep Hours (Beta_1 = {beta[0]:.4f}): Effect of an extra hour of sleep on assistance score.")
print(f"   - Activity Level (Beta_2 = {beta[1]:.4f}): Effect of an incremental unit increase in physical activity.")
print(f"   - Stress Level (Beta_3 = {beta[2]:.4f}): Effect of higher stress on the need for cognitive assistance.")

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print(f"\ng. Predicted Assistance Score for new user {new_user}: {predicted_score:.2f}")
```

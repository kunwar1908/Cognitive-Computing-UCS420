```python
roll_number=1024170439
l=[]
while roll_number>0:
    tba=roll_number%10
    roll_number=roll_number//10
    l.append(tba*10)
l.reverse()
print(l)
```

```python
#(i)
print(l)
```

```python
#(ii)
l.append(30)
print(l)
l.insert(2,80)
print(l)
```

```python
#(iii)
l.pop()
print(l)
l.remove(80)
print(l)
```

```python
#(iv)
l2=[i for i in l]
l2.sort()
print(l2)
l2.sort(reverse=True)
print(l2)
```

```python
#(v)
print(l[0:3])
print(l[-1:-4:-1])
```

```python
#(vi)
avg_l=sum(l)/len(l)
new_l=[i for i in l if i>avg_l]
print(new_l)
```

```python
scores=tuple(l[0:8])
print(scores)
```

```python
#(i)
max_score=max(scores)
min_score=min(scores)
print("maximum score is",max_score)
print("maximum score index is",scores.index(max_score))
print("minimum score is",min_score)
print("frequency of minimum score is",scores.count(min_score))
```

```python
#(ii)
l_scores=list(scores)
l_scores=l_scores[-1::-1]
print(l_scores)
```

```python
print("scores tupple is",scores)
num=int(input("Enter a score"))
if(num in scores):
    print(scores.index(num))

else:
    print("not present")
```

```python
my_dict["age"]=my_dict["age"]+1;
print(my_dict)
```

```python
del_del=my_dict.copy()
del(del_del["branch"])
print(del_del)
```

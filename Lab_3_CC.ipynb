```python
import pandas as pd

# Create and save initial employee dataset
emp_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

df_emp = pd.DataFrame(emp_data)
df_emp.to_csv("employees.csv", index=False)

# Load the dataset
df = pd.read_csv("employees.csv")

# a) Shape of DataFrame
print("--- (a) Shape (Rows, Columns) ---")
print(df.shape)

# b) Summary info (Data types & non-null counts)
print("\n--- (b) DataFrame Info ---")
df.info()

# c) Descriptive statistics
print("\n--- (c) Descriptive Statistics ---")
print(df.describe())

# d) First 5 rows and last 3 rows
print("\n--- (d) First 5 Rows ---")
print(df.head(5))
print("\n--- (d) Last 3 Rows ---")
print(df.tail(3))

# e) Specific statistics
avg_salary = df["Salary"].mean()
total_bonus = df["Bonus"].sum()
min_age = df["Age"].min()
max_rating = df["Rating"].max()

print("\n--- (e) Summary Metrics ---")
print(f"i.   Average Salary: {avg_salary}")
print(f"ii.  Total Bonus Paid: {total_bonus}")
print(f"iii. Youngest Employee Age: {min_age}")
print(f"iv.  Highest Performance Rating: {max_rating}")

# f) Sort by Salary in descending order
print("\n--- (f) Sorted by Salary (Descending) ---")
print(df.sort_values(by="Salary", ascending=False))

# g) Categorize employees based on Rating
def categorize_rating(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

df["Performance_Category"] = df["Rating"].apply(categorize_rating)
print("\n--- (g) Categorized Performance Ratings ---")
print(df[["Name", "Rating", "Performance_Category"]])

# h) Identify missing values
print("\n--- (h) Missing Values Count ---")
print(df.isnull().sum())

# i) Rename Employee_ID to ID
df.rename(columns={"Employee_ID": "ID"}, inplace=True)
print("\n--- (i) Renamed Column ---")
print(df.columns)

# j) Filter conditions
print("\n--- (j) Employees with Experience > 5 AND in IT Department ---")
filtered_emp = df[(df["Years_of_Experience"] > 5) & (df["Department"] == "IT")]
print(filtered_emp)

# k) Add Tax column (10% of Salary)
df["Tax"] = df["Salary"] * 0.10
print("\n--- (k) Dataset with Tax Column ---")
print(df[["ID", "Name", "Salary", "Tax"]])

# l) Save modified DataFrame to a new CSV file
df.to_csv("modified_employees.csv", index=False)
print("\n--- (l) Saved modified DataFrame to 'modified_employees.csv' successfully. ---")
```

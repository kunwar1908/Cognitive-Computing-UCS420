```python
# Q1: Find maximum of two numbers
num1 = int(input("Enter number 1: "))
num2 = int(input("Enter number 2: "))

if num1 > num2:
    print(f"{num1} is greater than {num2}")
else:
    print(f"{num2} is greater than {num1}")
```

```python
# Q2: Check if a number is prime
num = int(input("Enter your number: "))
isPrime = True

for i in range(2, (num // 2) + 1):
    if num % i == 0:
        isPrime = False
        break

if isPrime and num > 1:
    print("Prime")
else:
    print("Not Prime")
```

```python
# Find numbers divisible by both 5 and 7 up to N and calculate their sum
num = int(input("Enter your number: "))
cumsum = 0

for i in range(1, num + 1):
    if (i % 7 == 0) and (i % 5 == 0):
        print(i)
        cumsum = cumsum + i

print("Sum:", cumsum)
```

```python
# Function to check prime status and sum all prime numbers up to N
def isPrime(num):
    if num < 2:
        return False
    for i in range(2, (num // 2) + 1):
        if num % i == 0:
            return False
    return True

cumsum = 0
num = int(input("Enter number: "))

for i in range(1, num + 1):
    if isPrime(i):
        cumsum += i

print("Sum of prime numbers:", cumsum)
```

### Q1: Basic Operations on 1D Array
- Create a 1D NumPy array with 5 elements[cite: 2].
- a) Add 2 to all elements[cite: 2].
- b) Multiply all elements by 3[cite: 2].
- c) Divide every element by 2[cite: 2].

```python
import numpy as np

# Create 1D array with 5 elements
arr1 = np.array([10, 20, 30, 40, 50])
print("Original Array:", arr1)

# a) Addition of 2 in all elements
print("a) After adding 2:", arr1 + 2)

# b) Multiply 3 with all elements
print("b) After multiplying by 3:", arr1 * 3)

# c) Divide every element by 2
print("c) After dividing by 2:", arr1 / 2)
```

### Q2: Questions on Basic NumPy Array
- a) Reverse `arr = np.array([1, 2, 3, 6, 4, 5])`[cite: 2].
- b) Find the most frequent value and its index/indices for `x` and `y`[cite: 2].

```python
# a) Reverse array
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print("a) Reversed Array:", reversed_arr)

# b) Function to find most frequent value(s) and their indices
def find_mode_and_indices(a, name="array"):
    unique_vals, counts = np.unique(a, return_counts=True)
    max_count = np.max(counts)
    most_frequent_vals = unique_vals[counts == max_count]

    print(f"\nResults for {name}:")
    print(f"  Most frequent value(s): {most_frequent_vals} (Frequency: {max_count})")
    for val in most_frequent_vals:
        indices = np.where(a == val)[0]
        print(f"  Indices of value {val}: {indices.tolist()}")

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

find_mode_and_indices(x, "x")
find_mode_and_indices(y, "y")
```

### Q4: Array Attributes, linspace, and Transpose
- Create 1D array of 25 numbers from 10 to 100 using `linspace()`[cite: 2].
- Print dimensions, shape, total elements, dtype, and total bytes consumed[cite: 2].
- Find transpose using `reshape()` and discuss using `.T`[cite: 2].

```python
# 1D array named Kunwarpreet (replace with your name if needed)
Kunwarpreet = np.linspace(10, 100, 25)
print("Array Kunwarpreet:\n", Kunwarpreet)

# Array attributes
print("\n--- Array Properties ---")
print("Dimensions (ndim):", Kunwarpreet.ndim)
print("Shape:", Kunwarpreet.shape)
print("Total elements (size):", Kunwarpreet.size)
print("Data type (dtype):", Kunwarpreet.dtype)
print("Total bytes consumed (nbytes):", Kunwarpreet.nbytes, "bytes")

# Transpose using reshape()
# A 1D array shape (25,) must be converted to a 2D shape (1, 25) to transpose to a column vector (25, 1)
transpose_via_reshape = Kunwarpreet.reshape(-1, 1)
print("\nTranspose via reshape(-1, 1) Shape:", transpose_via_reshape.shape)

# Transpose using .T attribute on 1D vs 2D
print("1D Array .T Shape (remains unchanged):", Kunwarpreet.T.shape)
print("2D Reshaped Array .T Shape:", Kunwarpreet.reshape(1, -1).T.shape)
print("\nExplanation: For a strictly 1D array, .T has no effect because there is only 1 axis. To obtain a column vector transpose, we must reshape it to 2D first.")
```

### Q5: 2-D Array Statistics, Reshaping, and Resizing
- Create 3x4 array with given values[cite: 2].
- Compute mean, median, max, min, and unique elements[cite: 2].
- Reshape to 4x3 (`reshaped_ucs420_<name>`)[cite: 2].
- Resize to 2x3 (`resized_ucs420_<name>`)[cite: 2].

```python
# Create 3x4 array
values = [10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]
ucs420_Kunwarpreet = np.array(values).reshape(3, 4)

print("Original 3x4 Array (ucs420_Kunwarpreet):\n", ucs420_Kunwarpreet)

# Statistical operations
print("\n--- Statistical Measures ---")
print("Mean:", np.mean(ucs420_Kunwarpreet))
print("Median:", np.median(ucs420_Kunwarpreet))
print("Max:", np.max(ucs420_Kunwarpreet))
print("Min:", np.min(ucs420_Kunwarpreet))
print("Unique elements:", np.unique(ucs420_Kunwarpreet))

# Reshape to 4 rows and 3 columns
reshaped_ucs420_Kunwarpreet = ucs420_Kunwarpreet.reshape(4, 3)
print("\nReshaped Array (4x3):\n", reshaped_ucs420_Kunwarpreet)

# Resize the array to 2 rows and 3 columns (takes the first 6 elements)
resized_ucs420_Kunwarpreet = np.resize(ucs420_Kunwarpreet, (2, 3))
print("\nResized Array (2x3):\n", resized_ucs420_Kunwarpreet)
```
